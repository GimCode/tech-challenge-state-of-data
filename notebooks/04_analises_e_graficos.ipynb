```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum as spark_sum, round as spark_round, desc
import matplotlib.pyplot as plt

spark = SparkSession.builder.appName("StateOfData-Analises").getOrCreate()

BUCKET = "tech-challenge-state-of-data"
GOLD_PATH = f"s3://{BUCKET}/gold/state_of_data"

resumo = spark.read.parquet(f"{GOLD_PATH}/resumo_pesquisa")
perfil = spark.read.parquet(f"{GOLD_PATH}/perfil_profissionais")
remuneracao = spark.read.parquet(f"{GOLD_PATH}/remuneracao")
mercado = spark.read.parquet(f"{GOLD_PATH}/mercado_trabalho")
tecnologias = spark.read.parquet(f"{GOLD_PATH}/tecnologias")
ia = spark.read.parquet(f"{GOLD_PATH}/inteligencia_artificial")
diversidade = spark.read.parquet(f"{GOLD_PATH}/diversidade")
```

```python
resumo_pd = resumo.orderBy("periodo_pesquisa").toPandas()
resumo_pd
```

```python
perfil_nivel = (
    perfil.groupBy("periodo_pesquisa", "nivel_carreira")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

perfil_nivel.show(50, truncate=False)
```

```python
perfil_nivel_pd = perfil_nivel.toPandas()
for periodo in perfil_nivel_pd["periodo_pesquisa"].unique():
    dados = perfil_nivel_pd[perfil_nivel_pd["periodo_pesquisa"] == periodo]
    dados.plot.bar(x="nivel_carreira", y="quantidade_profissionais", legend=False)
    plt.title(f"Profissionais por senioridade — {periodo}")
    plt.xlabel("Nível")
    plt.ylabel("Quantidade")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
diversidade_genero = (
    diversidade.groupBy("periodo_pesquisa", "genero")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

diversidade_genero.show(50, truncate=False)
```

```python
ia_uso = (
    ia.groupBy("periodo_pesquisa", "uso_ia_trabalho")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

ia_uso.show(50, truncate=False)
```

```python
clouds = (
    tecnologias.groupBy("periodo_pesquisa", "cloud_preferida")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

clouds.show(50, truncate=False)
```

```python
salarios = (
    remuneracao.groupBy("periodo_pesquisa", "faixa_salarial")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

salarios.show(100, truncate=False)
```

```python
regioes = (
    perfil.groupBy("periodo_pesquisa", "regiao")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

regioes.show(50, truncate=False)
```

```python
modelos = (
    perfil.groupBy("periodo_pesquisa", "modelo_trabalho")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

modelos.show(50, truncate=False)
```

```python
oportunidades = (
    mercado.groupBy("periodo_pesquisa", "oportunidade_buscada")
    .agg(spark_sum("quantidade_profissionais").alias("quantidade_profissionais"))
    .orderBy("periodo_pesquisa", desc("quantidade_profissionais"))
)

oportunidades.show(100, truncate=False)
```

As análises podem ser usadas no storytelling executivo para comparar evolução temporal, senioridade, remuneração, diversidade, adoção de cloud e IA, regiões, modelos de trabalho e oportunidades buscadas.
