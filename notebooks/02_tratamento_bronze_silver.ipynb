```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lit, trim
from pyspark.sql.types import StringType, IntegerType
```

```python
spark = SparkSession.builder \
    .appName("StateOfData-BronzeToSilver") \
    .getOrCreate()
```

```python
BUCKET = "tech-challenge-state-of-data"

BRONZE_PATH = f"s3://{BUCKET}/bronze/state_of_data"
SILVER_PATH = f"s3://{BUCKET}/silver/state_of_data/dados_harmonizados"
```

```python
df_2022 = spark.read.parquet(f"{BRONZE_PATH}/2022")
df_2024 = spark.read.parquet(f"{BRONZE_PATH}/2024")
df_2025 = spark.read.parquet(f"{BRONZE_PATH}/2025_2026")
```

```python
def get_column(df, candidates, default=None):
    for candidate in candidates:
        if candidate in df.columns:
            return col(candidate)
    return lit(default)
```

```python
silver_2022 = df_2022.select(
    get_column(df_2022, ["('P0', 'id')"]).alias("id_resposta"),
    get_column(df_2022, ["('P1_a ', 'Idade')"]).alias("idade"),
    get_column(df_2022, ["('P1_a_1 ', 'Faixa idade')"]).alias("faixa_idade"),
    get_column(df_2022, ["('P1_b ', 'Genero')"]).alias("genero"),
    get_column(df_2022, ["('P1_i_1 ', 'uf onde mora')"]).alias("uf"),
    get_column(df_2022, ["('P1_i_2 ', 'Regiao onde mora')"]).alias("regiao"),
    get_column(df_2022, ["('P1_l ', 'Nivel de Ensino')"]).alias("nivel_ensino"),
    get_column(df_2022, ["('P2_f ', 'Cargo Atual')"]).alias("cargo_atual"),
    get_column(df_2022, ["('P2_g ', 'Nivel')"]).alias("nivel_carreira"),
    get_column(df_2022, ["('P2_h ', 'Faixa salarial')"]).alias("faixa_salarial"),
    get_column(df_2022, ["('P2_i ', 'Quanto tempo de experiência na área de dados você tem?')"]).alias("experiencia_dados"),
    get_column(df_2022, ["('P2_j ', 'Quanto tempo de experiência na área de TI/Engenharia de Software você teve antes de começar a trabalhar na área de dados?')"]).alias("experiencia_ti"),
    get_column(df_2022, ["ano_pesquisa"]),
    get_column(df_2022, ["periodo_pesquisa"]),
    get_column(df_2022, ["arquivo_origem"]),
    get_column(df_2022, ["data_ingestao"])
)
```

```python
silver_2024 = df_2024.select(
    get_column(df_2024, ["id", "1.a_id", "0.a_id"]).alias("id_resposta"),
    get_column(df_2024, ["1.a_idade"]).alias("idade"),
    get_column(df_2024, ["1.a.1_faixa_idade"]).alias("faixa_idade"),
    get_column(df_2024, ["1.b_genero"]).alias("genero"),
    get_column(df_2024, ["1.i.1_uf_onde_mora"]).alias("uf"),
    get_column(df_2024, ["1.i.2_regiao_onde_mora"]).alias("regiao"),
    get_column(df_2024, ["1.l_nivel_de_ensino"]).alias("nivel_ensino"),
    get_column(df_2024, ["2.f_cargo_atual"]).alias("cargo_atual"),
    get_column(df_2024, ["2.g_nivel"]).alias("nivel_carreira"),
    get_column(df_2024, ["2.h_faixa_salarial"]).alias("faixa_salarial"),
    get_column(df_2024, ["2.i_tempo_de_experiencia_em_dados"]).alias("experiencia_dados"),
    get_column(df_2024, ["2.j_tempo_de_experiencia_em_ti"]).alias("experiencia_ti"),
    get_column(df_2024, ["2.r_modelo_de_trabalho_atual"]).alias("modelo_trabalho"),
    get_column(df_2024, ["4.h_cloud_(dia_a_dia)"]).alias("cloud_dia_a_dia"),
    get_column(df_2024, ["4.i_cloud_preferida"]).alias("cloud_preferida"),
    get_column(df_2024, ["4.m_usa_chatgpt_ou_copilot_no_trabalho?"]).alias("uso_ia_trabalho"),
    get_column(df_2024, ["5.b_oportunidade_buscada"]).alias("oportunidade_buscada"),
    get_column(df_2024, ["ano_pesquisa"]),
    get_column(df_2024, ["periodo_pesquisa"]),
    get_column(df_2024, ["arquivo_origem"]),
    get_column(df_2024, ["data_ingestao"])
)
```

```python
silver_2025 = df_2025.select(
    get_column(df_2025, ["id", "1.a_id"]).alias("id_resposta"),
    get_column(df_2025, ["1.a_idade"]).alias("idade"),
    get_column(df_2025, ["1.a.1_faixa_idade"]).alias("faixa_idade"),
    get_column(df_2025, ["1.b_genero"]).alias("genero"),
    get_column(df_2025, ["1.i.1_uf_onde_mora"]).alias("uf"),
    get_column(df_2025, ["1.i.2_regiao_onde_mora"]).alias("regiao"),
    get_column(df_2025, ["1.l_nivel_de_ensino"]).alias("nivel_ensino"),
    get_column(df_2025, ["2.f_cargo_atual"]).alias("cargo_atual"),
    get_column(df_2025, ["2.g_nivel"]).alias("nivel_carreira"),
    get_column(df_2025, ["2.h_faixa_salarial"]).alias("faixa_salarial"),
    get_column(df_2025, ["2.i_tempo_de_experiencia_em_dados"]).alias("experiencia_dados"),
    get_column(df_2025, ["2.j_tempo_de_experiencia_em_ti"]).alias("experiencia_ti"),
    get_column(df_2025, ["2.q_modelo_de_trabalho_atual"]).alias("modelo_trabalho"),
    get_column(df_2025, ["4.e_cloud_(dia_a_dia)"]).alias("cloud_dia_a_dia"),
    get_column(df_2025, ["4.f_cloud_preferida"]).alias("cloud_preferida"),
    get_column(df_2025, ["4.j_usa_chatgpt_ou_copilot_no_trabalho?"]).alias("uso_ia_trabalho"),
    get_column(df_2025, ["5.b_oportunidade_buscada"]).alias("oportunidade_buscada"),
    get_column(df_2025, ["ano_pesquisa"]),
    get_column(df_2025, ["periodo_pesquisa"]),
    get_column(df_2025, ["arquivo_origem"]),
    get_column(df_2025, ["data_ingestao"])
)
```

```python
silver = silver_2022.unionByName(
    silver_2024,
    allowMissingColumns=True
).unionByName(
    silver_2025,
    allowMissingColumns=True
)
```

```python
silver = silver     .withColumn("id_resposta", trim(col("id_resposta").cast(StringType())))     .withColumn("idade", col("idade").cast(IntegerType()))     .withColumn("faixa_idade", trim(col("faixa_idade").cast(StringType())))     .withColumn("genero", trim(col("genero").cast(StringType())))     .withColumn("uf", trim(col("uf").cast(StringType())))     .withColumn("regiao", trim(col("regiao").cast(StringType())))     .withColumn("nivel_ensino", trim(col("nivel_ensino").cast(StringType())))     .withColumn("cargo_atual", trim(col("cargo_atual").cast(StringType())))     .withColumn("nivel_carreira", trim(col("nivel_carreira").cast(StringType())))     .withColumn("faixa_salarial", trim(col("faixa_salarial").cast(StringType())))     .withColumn("experiencia_dados", trim(col("experiencia_dados").cast(StringType())))     .withColumn("experiencia_ti", trim(col("experiencia_ti").cast(StringType())))     .withColumn("modelo_trabalho", trim(col("modelo_trabalho").cast(StringType())))     .withColumn("cloud_dia_a_dia", trim(col("cloud_dia_a_dia").cast(StringType())))     .withColumn("cloud_preferida", trim(col("cloud_preferida").cast(StringType())))     .withColumn("uso_ia_trabalho", trim(col("uso_ia_trabalho").cast(StringType())))     .withColumn("oportunidade_buscada", trim(col("oportunidade_buscada").cast(StringType())))
```

```python
silver = silver.dropDuplicates(["id_resposta", "periodo_pesquisa"])
```

```python
silver.select(
    "periodo_pesquisa",
    "id_resposta",
    "genero",
    "regiao",
    "nivel_ensino",
    "cargo_atual",
    "nivel_carreira",
    "faixa_salarial",
    "experiencia_dados",
    "modelo_trabalho",
    "uso_ia_trabalho"
).show(20, truncate=False)
```

```python
silver.groupBy("periodo_pesquisa")     .count()     .orderBy("periodo_pesquisa")     .show()
```

```python
silver.groupBy("periodo_pesquisa", "genero")     .count()     .orderBy("periodo_pesquisa", "genero")     .show()
```

```python
silver.groupBy("periodo_pesquisa", "regiao")     .count()     .orderBy("periodo_pesquisa", "regiao")     .show()
```

```python
silver.write     .mode("overwrite")     .partitionBy("periodo_pesquisa")     .parquet(SILVER_PATH)
```

```python
silver_validacao = spark.read.parquet(SILVER_PATH)

print(f"Total de registros Silver: {silver_validacao.count()}")
silver_validacao.printSchema()
```
